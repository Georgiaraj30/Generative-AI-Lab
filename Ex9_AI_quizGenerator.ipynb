```python

import gradio as gr
from google import genai
from getpass import getpass

# Enter your NEW Gemini API key
API_KEY = getpass("Enter your Gemini API key: ")

client = genai.Client(api_key=API_KEY)

def generate_quiz(topic):
    if not topic.strip():
        return "Please enter a topic first."

    try:
        prompt = f"""
        Create a short quiz about {topic}.

        Give exactly:
        3 MCQ questions with 4 options each and answers.
        2 descriptive questions with short answers.

        Keep everything short and clear.
        """

        response = client.models.generate_content(
            model="gemini-3.6-flash",
            contents=prompt
        )

        if response.text:
            return response.text
        return "The model returned an empty response."

    except Exception as e:
        return f"Error: {str(e)}"


demo = gr.Interface(
    fn=generate_quiz,
    inputs=gr.Textbox(
        label="Enter Topic",
        placeholder="Example: Machine Learning"
    ),
    outputs=gr.Textbox(label="Generated Quiz", lines=20),
    title="AI Quiz Generator",
    description="Generate MCQs and descriptive questions using Generative AI."
)

demo.launch()
```
